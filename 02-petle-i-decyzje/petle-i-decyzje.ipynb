## Ściąga: decyzje i pętle

Kolumna Kotlina pochodzi z dokumentacji języka — porównaj ją z tym, co już znasz.

| Kotlin | JavaScript | Python | Uwagi |
| --- | --- | --- | --- |
| `if (x > 0) { … } else if (x < 0) { … } else { … }` | tak samo jak w Kotlinie | `if x > 0:` … `elif x < 0:` … `else:` … | dwukropek i wcięcie zamiast klamer |
| `x in 1..6` | `x >= 1 && x <= 6` | `1 <= x <= 6` | porównania można łączyć w łańcuch |
| `a && b`, `a \|\| b`, `!a` | tak samo | `a and b`, `a or b`, `not a` | |
| `when (x) { 1 -> …; 2, 3 -> …; else -> … }` | `switch (x) { case 1: …; break; default: … }` | `match x:` `case 1:` … `case 2 \| 3:` … `case _:` … | `match` jest od Pythona 3.10; `_` to „każda inna wartość” |
| `for (i in 1..10)` | `for (let i = 1; i <= 10; i++)` | `for i in range(1, 11):` | **koniec `range` się nie liczy** |
| `for (i in 10 downTo 1 step 2)` | `for (let i = 10; i >= 1; i -= 2)` | `for i in range(10, 0, -2):` | trzeci argument to krok |
| `for (x in lista)` | `for (const x of lista)` | `for x in lista:` | |
| `for ((i, x) in lista.withIndex())` | `lista.forEach((x, i) => …)` | `for i, x in enumerate(lista):` | `enumerate(lista, start=1)` numeruje od 1 |
| `for ((k, v) in mapa)` | `for (const [k, v] of Object.entries(obiekt))` | `for k, v in slownik.items():` | |
| `while (w) { … }` | tak samo | `while w:` | |
| `do { … } while (w)` | tak samo | nie ma; `while True:` … `if not w: break` | |
| `break`, `continue` | tak samo | `break`, `continue` | `break` kończy pętlę, `continue` przechodzi do następnego obrotu |
| — | — | `for …:` … `else:` … | `else` przy pętli wykona się, gdy pętla skończyła się **bez** `break` |
| pusty blok `{ }` | `{ }` | `pass` | w Pythonie blok nie może być pusty; `pass` znaczy „nic nie rób” |
| `print("a")` (bez nowej linii) | `process.stdout.write("a")` | `print("a", end="")` | `end` zastępuje znak nowej linii na końcu |

Uruchom przykład i porównaj z tabelą:

````python
# enumerate: numer i element listy
for nr, przedmiot in enumerate(["polski", "matematyka", "informatyka"], start=1):
    print(nr, przedmiot)

# for ... else: else wykona się, gdy pętla nie trafiła na break
for liczba in [3, 7, 11]:
    if liczba % 2 == 0:
        print("Jest liczba parzysta:", liczba)
        break
else:
    print("Nie ma liczby parzystej")

# match: jak when w Kotlinie
dzien = "sob"
match dzien:
    case "sob" | "nd":
        print("weekend")
    case "pt":
        print("prawie weekend")
    case _:
        print("dzień szkolny")

# pass: pusty blok, np. funkcja do napisania później
if dzien == "pon":
    pass

# print bez nowej linii
for i in range(5):
    print(i, end=" ")
print()
````

## Zadanie 2. Napraw program: punkty na oceny

Program miał zamienić wyniki sprawdzianu (punkty na 40) na oceny i policzyć, ile osób zdało (ocena co najmniej 2). Progi: od 90% — 5, od 70% — 4, od 50% — 3, od 30% — 2, poniżej — 1. Natalia była nieobecna, więc zamiast punktów ma `None`.

Uruchom program. Wynik jest zły w dwóch miejscach: prawie wszyscy dostają 2, a lista urywa się na Natalii. Znajdź **dwa błędy** i popraw program tak, żeby wypisał oczekiwany wynik.

````python
# Napraw: w programie są dwa błędy
wyniki = {"Ola": 38, "Kuba": 22, "Natalia": None, "Bartek": 10, "Zuza": 32, "Igor": 14}
MAKS = 40
zdali = 0

for imie, punkty in wyniki.items():
    if punkty is None:
        print(f"{imie}: brak wyniku (nieobecność)")
        break
    procent = punkty / MAKS * 100
    if procent >= 30:
        ocena = 2
    elif procent >= 50:
        ocena = 3
    elif procent >= 70:
        ocena = 4
    elif procent >= 90:
        ocena = 5
    else:
        ocena = 1
    if ocena >= 2:
        zdali += 1
    print(f"{imie}: {punkty}/{MAKS} pkt ({procent:.0f}%) → ocena {ocena}")

print(f"Zdało: {zdali}")
````

**Oczekiwany wynik:**

```text
Ola: 38/40 pkt (95%) → ocena 5
Kuba: 22/40 pkt (55%) → ocena 3
Natalia: brak wyniku (nieobecność)
Bartek: 10/40 pkt (25%) → ocena 1
Zuza: 32/40 pkt (80%) → ocena 4
Igor: 14/40 pkt (35%) → ocena 2
Zdało: 4
```
