## Ściąga: funkcje

Kolumna Kotlina pochodzi z dokumentacji języka.

| Kotlin | JavaScript | Python | Uwagi |
| --- | --- | --- | --- |
| `fun pole(a: Int, b: Int): Int { return a * b }` | `function pole(a, b) { return a * b }` | `def pole(a, b):` i w nowej linii, z wcięciem, `return a * b` | bez typów i bez klamer |
| `fun pasek(w: Int, maks: Int = 10)` | `function pasek(w, maks = 10)` | `def pasek(w, maks=10):` | parametr z wartością domyślną; parametry domyślne na końcu |
| `pasek(7, maks = 20)` | — (brak argumentów nazwanych) | `pasek(7, maks=20)` | argument nazwany |
| `fun suma(vararg liczby: Int)` | `function suma(...liczby)` | `def suma(*liczby):` | dowolnie wiele argumentów; w środku `liczby` to krotka |
| `suma(*tablica)` | `suma(...tablica)` | `suma(*lista)` | gwiazdka przy wywołaniu „rozpakowuje” listę na argumenty |
| funkcja bez `return` zwraca `Unit` | zwraca `undefined` | zwraca `None` | |
| `{ x -> x * 2 }` | `x => x * 2` | `lambda x: x * 2` | lambda: jedno wyrażenie, bez słowa `return` |
| `lista.sortedBy { it.nazwisko }` | `[...lista].sort((a, b) => a.nazwisko.localeCompare(b.nazwisko))` | `sorted(lista, key=lambda u: u["nazwisko"])` | `sorted` zwraca nową listę; `lista.sort(key=…)` sortuje w miejscu |
| `lista.sortedByDescending { it.srednia }` | `[...lista].sort((a, b) => b.srednia - a.srednia)` | `sorted(lista, key=lambda u: u["srednia"], reverse=True)` | |
| `val` / `var` w funkcji — zmienna lokalna | `let` / `const` w bloku `{ }` — lokalna w bloku | zmienna, do której **przypisujesz** wartość w funkcji, jest lokalna w całej funkcji | `if` i `for` w Pythonie **nie** tworzą osobnego zasięgu |

Uruchom przykład:

````python
def przywitaj(imie, powitanie="Cześć"):
    """Zwraca powitanie dla jednej osoby."""
    return f"{powitanie}, {imie}!"


def suma(*liczby):
    print("  liczby to krotka:", liczby)
    wynik = 0
    for x in liczby:
        wynik += x
    return wynik


print(przywitaj("Ola"))
print(przywitaj("Kuba", powitanie="Dzień dobry"))
print(suma(1, 2, 3))
punkty = [10, 20, 5]
print(suma(*punkty))
podwoj = lambda x: x * 2
print(podwoj(21))
print(sorted(["Zuza", "Ola", "Bartek"], key=lambda imie: len(imie)))
````

## Zasięg zmiennych: lokalne i globalne

Zmienna utworzona poza funkcjami jest **globalna** — każda funkcja może ją odczytać. Ale gdy funkcja **przypisuje** zmiennej wartość (`=`, `+=`), Python uznaje ją za **lokalną** w całej tej funkcji: to nowa zmienna, która znika po zakończeniu funkcji, a globalna o tej samej nazwie się nie zmienia.

W Kotlinie i JavaScripcie jest podobnie, ale z jedną różnicą: tam `val` / `let` w bloku `if` albo pętli tworzy zmienną widoczną tylko w tym bloku. W Pythonie zasięg tworzą tylko funkcje — zmienna z pętli `for` istnieje także po pętli.

Uruchom przykład:

````python
poziom = 1


def pokaz_poziom():
    print(f"Odczyt zmiennej globalnej: {poziom}")


def zmien_poziom():
    poziom = 99          # przypisanie: to nowa zmienna LOKALNA
    print(f"Poziom w funkcji: {poziom}")


pokaz_poziom()
zmien_poziom()
print(f"Poziom po wywołaniu funkcji: {poziom}")

for i in range(3):
    pass
print(f"Zmienna i istnieje po pętli: {i}")
````

Program miał dodawać punkty graczowi, ale kończy się błędem `UnboundLocalError`. Uruchom go i przeczytaj komunikat: funkcja przypisuje wartość zmiennej `punkty`, więc dla Pythona to zmienna lokalna — a lokalna nie ma jeszcze wartości, gdy próbujemy ją odczytać po prawej stronie `=`.

Napraw program **bez słowa `global`**: funkcja `dodaj_punkty(punkty, ile)` dostaje obecne punkty jako parametr i **zwraca** nowe, a program zapisuje wynik: `punkty = dodaj_punkty(punkty, 10)`.

````python
# Napraw: program kończy się błędem UnboundLocalError
punkty = 0


def dodaj_punkty(ile):
    punkty = punkty + ile
    print(f"+{ile} pkt, razem: {punkty}")


dodaj_punkty(10)
dodaj_punkty(5)
dodaj_punkty(20)
print(f"Wynik końcowy: {punkty}")
````

**Oczekiwany wynik:**

```text
+10 pkt, razem: 10
+5 pkt, razem: 15
+20 pkt, razem: 35
Wynik końcowy: 35
```

Słowo `global punkty` na początku funkcji też by „naprawiło” błąd, ale wtedy funkcja po cichu zmienia coś poza sobą. Przy dwóch graczach potrzebne byłyby dwie funkcje albo kolejne zmienne globalne, a wyniku funkcji nie da się sprawdzić bez zaglądania do całego programu.
````python
oceny_3a = [5, 4, 0, 6, 3, 5]
oceny_3b = [2, 3, 3, 7, 4, 2, 3]
oceny_3c = [6, 5, 5, 4, -1, 6]

# klasa 3A
suma = 0
ile = 0
for ocena in oceny_3a:
    if ocena < 1 or ocena > 6:
        continue
    suma += ocena
    ile += 1
sr = suma / ile
if sr >= 5.5:
    slownie = "celujący"
elif sr >= 4.5:
    slownie = "bardzo dobry"
elif sr >= 3.5:
    slownie = "dobry"
elif sr >= 2.5:
    slownie = "dostateczny"
elif sr >= 1.5:
    slownie = "dopuszczający"
else:
    slownie = "niedostateczny"
print(f"3A: średnia {sr:.2f} ({slownie}), ocen: {ile}")

# klasa 3B
suma = 0
ile = 0
for ocena in oceny_3b:
    if ocena < 1 or ocena > 6:
        continue
    suma += ocena
    ile += 1
sr = suma / ile
if sr >= 5.5:
    slownie = "celujący"
elif sr >= 4.5:
    slownie = "bardzo dobry"
elif sr >= 3.5:
    slownie = "dobry"
elif sr >= 2.5:
    slownie = "dostateczny"
elif sr >= 1.5:
    slownie = "dopuszczający"
else:
    slownie = "niedostateczny"
print(f"3B: średnia {sr:.2f} ({slownie}), ocen: {ile}")

# klasa 3C
suma = 0
for ocena in oceny_3c:
    if ocena < 1 or ocena > 6:
        continue
    suma += ocena
    ile += 1
sr = suma / ile
if sr >= 5.5:
    slownie = "celujący"
elif sr >= 4.5:
    slownie = "bardzo dobry"
elif sr >= 3.5:
    slownie = "dobry"
elif sr >= 2.5:
    slownie = "dostateczny"
elif sr >= 1.5:
    slownie = "dopuszczający"
else:
    slownie = "niedostateczny"
print(f"3C: średnia {sr:.2f} ({slownie}), ocen: {ile}")
````
